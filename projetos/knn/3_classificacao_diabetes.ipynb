### Este projeto tem como objetivo de predição de diabetes utilizando o algoritmo KNN.

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
import pandas as pd
```

```python
diabetes = load_diabetes()
diabetes_df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
diabetes_df['target'] = diabetes.target
diabetes_df.head()
```

#### Avaliar o modelo

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
r2 = modelo_knn.score(X, y)
print(f"R² do modelo KNN: {r2:.2f}")
```

```python
# RMSE
y_pred = modelo_knn.predict(X)
rmse = np.sqrt(mean_squared_error(y, y_pred))
print(f"RMSE do modelo KNN: {rmse:.2f}")
```

```python
# MAE
mae = mean_absolute_error(y, y_pred)
print(f"MAE do modelo KNN: {mae:.2f}")
```

#### após as validações eu entendo que o KNN nãop seria o melhor modelo para esse dataset, pois a acurácia e o R² estão baixos. O KNN é um algoritmo de classificação e regressão que pode ser sensível a outliers e pode não capturar bem as relações complexas entre as variáveis.
