#### Árvore de Decisão

## Resumo breve: Árvore de Decisão (em termos simples)

A **árvore de decisão** é um modelo que toma decisões em etapas, como um fluxo de perguntas de **sim/não**.
Ela vai dividindo os dados em grupos menores até chegar a uma resposta final (classe).

### Como ela decide a melhor pergunta?

Em cada etapa, a árvore testa várias perguntas e escolhe a que **melhor separa** os dados.
Para isso, ela mede a “bagunça” dos grupos. A medida mais usada é a **entropia**.

### Entropia (explicação didática)

Pense na entropia como o **nível de incerteza**:
- **Entropia alta**: grupo muito misturado (ex.: 50% de cada classe).
- **Entropia baixa**: grupo mais “puro” (quase tudo da mesma classe).
- **Entropia zero**: grupo totalmente puro (100% de uma classe).

Fórmula da entropia:

$$H(S)=-\sum_{i=1}^{k} p_i\log_2(p_i)$$

Onde:
- $p_i$ é a proporção de cada classe no grupo.
- Se as classes estão equilibradas, a entropia aumenta.
- Se uma classe domina, a entropia diminui.

### Ganho de informação (o cálculo principal da árvore)

A árvore escolhe a divisão que mais **reduz a entropia**.
Essa redução é o **ganho de informação**:

$$\text{Ganho}=H(\text{pai})-\sum_j \frac{|S_j|}{|S|}H(S_j)$$

Leitura simples da fórmula:
- Entropia antes da divisão
- menos a média ponderada das entropias depois da divisão
- resultado: quanto a divisão melhorou a organização dos dados.

### Exemplo simples

Problema: classificar fruta em **maçã** ou **laranja**.

Pergunta possível: “A cor é laranja?”
- Se sim, cai num grupo com maioria de laranjas.
- Se não, cai num grupo com maioria de maçãs.

Se essa pergunta deixa os grupos mais puros, ela tem alto ganho de informação e vira um nó da árvore.

### Exemplo do mundo real

Para prever risco de crédito, a árvore pode perguntar:
1. Renda > X?
2. Histórico de atraso?
3. Tempo de emprego?

Cada pergunta tenta reduzir a incerteza até chegar na decisão final.
